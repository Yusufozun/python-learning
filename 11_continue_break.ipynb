##30 a geldiğinde listeyi durdur 

```python
my_list =[10,20,30,40,50,60]
for num in my_list:
    if num == 30:
        break
    print(num*5)
```

##30 un 5 katını almasın atlayıp devam etsin dersek;

```python
for num in my_list:
    if num == 30:
        continue
    print(num*5)
```
