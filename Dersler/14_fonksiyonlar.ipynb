##arbitrary arguments && key word arguments

```python
def summation_2(*args):
    return sum(args)
    
```

```python
summation_2(10,20,30) 
```

```python
summation_2(70,60,50) 
```

```python
def my_func(*args):
    print(args)
```

```python
my_func(10,20)
```

```python
my_func("a","b",1,2)
```

```python
def my_func_2(*yusuf): ##Yani olay args da değil * da bunu anlatmak için yazılmıs bir kod 
    print(yusuf)
```
