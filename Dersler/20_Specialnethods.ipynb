# Normalde print ve len fonksiyonları çıktı vermiyo bu yüzden str ve len özel methodlarını kullanıyoruz

```python
class Fruits():
    def __init__(self,name,calories):
        self.name=name
        self.calories=calories
    def __str__(self):
        return f"{self.name} has {self.calories}"
    def __len__(self):
        return self.calories
```

```python
my_fruits=Fruits("Banana",200)
```

```python
my_fruits.calories
```

```python
my_fruits.name
```

```python
print(my_fruits)
```

```python
len(my_fruits)
```
