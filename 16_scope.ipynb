#LEGB
L -> Local
E -> enclosing
G -> Global
B -> Built-in
 


```python
my_string="Yusuf"
#Global
def my_func():
    my_string="james"
    #Enclosing
    def my_func_2():
        #Local
        my_string="Lars"
        print(my_string) 
    my_func_2()
```

```python
my_func()
```
