##RANGE

```python
range(4)
list(range(4))
for number in list(range(4)):
    print(number*5)
```

```python
for num in list(range (5,8)):
    print(num)
```

```python
for num in list(range (5,12,3)):
    print(num)
```

##enumerate

```python
index=0 
for number in list (range (0,4)):
    print(f"no: {number} ix: {index}")
    index+=1
```

```python
for number in enumerate(range(0,4)):
    print(number)
```

```python
for index ,number in enumerate(range (0,4)):
    print(number)
```

```python
for index ,number in enumerate(range (0,4)):
    print(index)
```
