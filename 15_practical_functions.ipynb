```python
def divide(number):
    return number/2
```

```python
divide(10)
```

```python
my_list=[1,2,3,4,5,6]
```

```python
for num in my_list:
    print(divide(num))
```

 ##Map:İki parametreliidir (fonksiyon,liste) yaptığı şey listendeki şeyleri fonksiyona uyarlayıp düzenlemek


```python
list(map(divide,my_list))
```

```python
def control_string(string):
    return "Metalica"in string

```

```python
control_string("Metalica aasfakjh")
```

```python
my_artist_list=["muuslum","madonna","quenn","Metalica","Metalica2"]
```

```python
list(map(control_string,my_artist_list))
```

#Filter:Mapdeki true ları gösteren filtre 

```python
list(filter(control_string,my_artist_list))
```

#Lambda

```python
def multiply(number):
    return number*5
```

```python
multiply(3)
```

```python
multiply=lambda number:number*5
```

```python
multiply(5)
```

```python
my_list_3=[3,5,7,9]
```

```python
list(map(lambda num:num*4,my_list_3))
```
