```python
def decorator_function(func):
    def wrapper_function():
        print("wrapper start")
        func()
        print("wrapper stopped")
    return wrapper_function
```

```python
def func_new():
    print("hello world")
```

```python
example_func=decorator_function(func_new)
```

```python
example_func()
```

#Decorator kullanımı ile bu kadar koddan kurtuluyoruz

```python
@decorator_function
def func_new():
    print("hello world")
```

```python
func_new()
```
